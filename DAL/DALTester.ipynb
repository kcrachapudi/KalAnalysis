```python
import DBLayer
import MockDBEngine
import pandas as pd

print('Getting DB Engine')
dbEngine:DBLayer.DBEngine = MockDBEngine.GetDBEngine()

print('Get all DB Tables')
DBTables = dbEngine.GetTableNames()
print(f'Printing DBTables')    
print(f'{DBTables}')    

print('Get Table Data - Brokerage')
dfBrokerageData = dbEngine.GetTableData('Brokerage')
print('Printing Brokerage Table Data')
print(dfBrokerageData)
```

```python
print('Get Watchlists')
wlists:pd.DataFrame = dbEngine.GetWatchlistNames()
print('Printing Watchlists')
print(wlists)

for wlist in wlists:
    print(f'Get Watchlist Details for {wlist}')
    edfWatchlistDetails = dbEngine.GetWatchlistSymbols(wlist) 
    print(edfWatchlistDetails)
```
