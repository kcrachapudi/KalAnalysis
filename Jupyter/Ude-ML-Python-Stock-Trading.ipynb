```python
import pandas as pd
import matplotlib.pyplot as plt
import datetime
from datetime import date, time

#Install Yahoo Finance Libraries
try:
    import yfinance
except:
    !pip install -q yfinance
    import yfinance
    
try:
    import yahoofinancials
except:
    !pip install -q yahoofinancials
    import yahoofinancials
    
import yfinance as yf
from yahoofinancials import YahooFinancials

import warnings
warnings.filterwarnings('ignore')

#Download Historical Data
startdate='2010-01-01'
enddate=date.today()
dfData:pd.DataFrame = yf.download('AAPL', start=startdate, end=enddate, progress=False)
```

```python
print(dfData.head())
#Remove Ticker row from Column Name
dfData = dfData.droplevel(1, axis=1)

# Rename the columns using the dictionary
rename_dict = {'Close': 'close', 'Volume': 'volume', 'High':'high', 'Open':'open', 'Low':'low'}
dfData.rename(columns=rename_dict, inplace=True)
```

```python
#Chart the Stock Performance
print(dfData.head())
dfData.plot(y='close', title='AAPL Stock Performance')
plt.show()
```

```python
#Create Technical Analysis Indicators and Plot 
from ta import add_all_ta_features
from ta.utils import dropna
from ta.volatility import BollingerBands
from ta.momentum import RSIIndicator
from ta.trend import CCIIndicator

# Add all ta features
dfData = add_all_ta_features(dfData, open="open", high="high", low="low", close="close", volume="volume")

# Calculate RSI
dfData['RSI-2'] = RSIIndicator(close=dfData['close'], window=14)
#View Data so far
print(dfData.tail())

dfData['CCI(30)'] = CCIIndicator(high=dfData['high'], low=dfData['low'], close=dfData['close'], window=30)

#View Data so far
print(dfData.tail())

#Drop NaN Values
dfData.dropna()

#View Data so far
print(dfData.tail())

#Plot showing some Indicators above
#dfData.plot(y='RSI(2)')
#fData.plot(y='CCI(100)')
```
