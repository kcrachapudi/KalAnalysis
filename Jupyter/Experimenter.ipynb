```python
import pandas as pd

# Create a DataFrame with MultiIndex columns
data = {'A': [1, 2], 'B': [3, 4]}
columns = pd.MultiIndex.from_tuples([('X', 'a'), ('Y', 'b')])
df = pd.DataFrame(data, columns=columns)
print(df)
# Drop the top level of the columns
df = df.droplevel(0, axis=1)
print(df)
```
